1. Đọc dữ liệu và kiểm tra cấu trúc

```python

import pandas as pd
import numpy as np
from pathlib import Path

# Xác định thư mục gốc dự án
ROOT = Path.cwd()

if ROOT.name == "notebooks":
    ROOT = ROOT.parent

DATA_PATH = ROOT / "data" / "raw" / "Online Retail.xlsx"

# Đọc dữ liệu Excel
df = pd.read_excel(DATA_PATH)

print("Số dòng:", df.shape[0])
print("Số cột:", df.shape[1])

print("\n5 dòng đầu tiên:")
display(df.head())

print("\nThông tin dữ liệu:")
df.info()
```

2. Kiểm tra Missing Values

```python

# CELL 2: MISSING VALUES ANALYSIS

# Đếm số giá trị bị thiếu trong từng cột
missing_count = df.isna().sum()

# Tính tỷ lệ phần trăm bị thiếu
missing_percent = (df.isna().mean() * 100).round(2)

# Tạo bảng tổng hợp
missing_table = pd.DataFrame({
    "Missing_Count": missing_count,
    "Missing_Percent (%)": missing_percent
})

# Sắp xếp theo số giá trị thiếu giảm dần
missing_table = missing_table.sort_values(
    by="Missing_Count",
    ascending=False
)

print("BẢNG KIỂM TRA MISSING VALUES")
display(missing_table)
```

Kiểm tra thêm các dòng bị thiếu

```python

# Xem 5 dòng thiếu CustomerID
print("Các dòng thiếu CustomerID:")
display(df[df["CustomerID"].isna()].head(5))

# Xem 5 dòng thiếu Description
print("Các dòng thiếu Description:")
display(df[df["Description"].isna()].head(5))
```

3: Kiểm tra dữ liệu trùng lặp

```python

# CELL 3: DUPLICATE ROWS ANALYSIS

# 1. Đếm số dòng trùng hoàn toàn
duplicate_count = df.duplicated().sum()

print("Số dòng trùng hoàn toàn:", duplicate_count)

# 2. Xác định tất cả các dòng thuộc nhóm trùng
duplicate_rows = df[
    df.duplicated(keep=False)
]

print("Số dòng thuộc các nhóm trùng:", len(duplicate_rows))

# 3. Hiển thị 10 dòng ví dụ
print("\nVí dụ các dòng trùng lặp:")
display(duplicate_rows.head(10))

# 4. Tính tỷ lệ dòng trùng
duplicate_percent = (
    duplicate_count / len(df) * 100
)

print(
    f"\nTỷ lệ dòng trùng: {duplicate_percent:.2f}%"
)
```

4.Kiểm tra hóa đơn hủy (Cancelled Invoices)

```python

# CELL 4: CANCELLED INVOICES ANALYSIS

# 1. Chuyển InvoiceNo sang dạng chuỗi
invoice_text = (
    df["InvoiceNo"]
    .astype("string")
    .str.strip()
    .str.upper()
)

# 2. Xác định mã hóa đơn bắt đầu bằng C
cancelled_mask = invoice_text.str.startswith(
    "C", na=False
)

# 3. Lấy các dòng thuộc hóa đơn hủy
cancelled_rows = df[cancelled_mask]

# 4. Đếm số dòng và số hóa đơn duy nhất
cancelled_count = len(cancelled_rows)

cancelled_invoices = (
    invoice_text[cancelled_mask].nunique()
)

# 5. Tính tỷ lệ
cancelled_percent = (
    cancelled_count / len(df) * 100
)

print("Số dòng thuộc hóa đơn hủy:", cancelled_count)
print("Số hóa đơn hủy duy nhất:", cancelled_invoices)
print(f"Tỷ lệ dòng hủy: {cancelled_percent:.2f}%")

# 6. Hiển thị 10 dòng ví dụ
print("\nVí dụ các giao dịch hủy:")
display(cancelled_rows.head(10))
```

Bộ dữ liệu ghi nhận 9.288 dòng thuộc 3.836 hóa đơn hủy, chiếm 1,71% tổng số dòng giao dịch. Các giao dịch này thường có số lượng âm và có thể liên quan đến hủy, trả hàng hoặc điều chỉnh giao dịch. Do đó, cần phân loại riêng và không đưa vào chỉ số Gross Sales của tập bán hàng hợp lệ.

5: Kiểm tra Quantity và UnitPrice

```python

# CELL 5: QUANTITY AND UNIT PRICE ANALYSIS

# 1. Chuyển dữ liệu sang kiểu số
quantity = pd.to_numeric(
    df["Quantity"], errors="coerce"
)

price = pd.to_numeric(
    df["UnitPrice"], errors="coerce"
)

# 2. Kiểm tra Quantity
print("===== QUANTITY ANALYSIS =====")

print("Quantity < 0:", (quantity < 0).sum())
print("Quantity = 0:", (quantity == 0).sum())
print("Quantity > 0:", (quantity > 0).sum())
print("Quantity <= 0:", (quantity <= 0).sum())

print("\n===== UNIT PRICE ANALYSIS =====")

# 3. Kiểm tra UnitPrice
print("UnitPrice < 0:", (price < 0).sum())
print("UnitPrice = 0:", (price == 0).sum())
print("UnitPrice > 0:", (price > 0).sum())
print("UnitPrice <= 0:", (price <= 0).sum())

# 4. Thống kê phân phối
print("\n===== DESCRIPTIVE STATISTICS =====")

display(df[["Quantity", "UnitPrice"]].describe())

# 5. Xem 10 dòng Quantity âm
print("\n===== NEGATIVE QUANTITY EXAMPLES =====")

display(df[quantity < 0].head(10))

# 6. Xem 10 dòng UnitPrice bằng 0 hoặc âm
print("\n===== NON-POSITIVE PRICE EXAMPLES =====")

display(df[price <= 0].head(10))
```

Giá trị trung bình là 9,55, cao hơn đáng kể so với trung vị là 3, cho thấy phân phối số lượng có thể chịu ảnh hưởng bởi các giao dịch có giá trị lớn bất thường. Do đó, cần kiểm tra và phân loại những giao dịch này trước khi phân tích doanh thu.

```python

# CELL 6: INVOICE DATE ANALYSIS

# 1. Chuyển cột InvoiceDate sang kiểu ngày giờ
invoice_dates = pd.to_datetime(
    df["InvoiceDate"],
    errors="coerce"
)

# 2. Kiểm tra phạm vi thời gian
print("===== DATE RANGE =====")
print("Ngày bắt đầu:", invoice_dates.min())
print("Ngày kết thúc:", invoice_dates.max())

# 3. Kiểm tra ngày thiếu hoặc không hợp lệ
print("\n===== INVALID DATES =====")
invalid_dates = invoice_dates.isna().sum()
print("Số dòng ngày không hợp lệ:", invalid_dates)

# 4. Thống kê số dòng theo năm
print("\n===== TRANSACTIONS BY YEAR =====")

year_counts = (
    invoice_dates.dt.year
    .value_counts()
    .sort_index()
)

display(year_counts)

# 5. Thống kê số dòng theo tháng
print("\n===== TRANSACTIONS BY MONTH =====")

month_counts = (
    invoice_dates.dt.to_period("M")
    .value_counts()
    .sort_index()
)

display(month_counts)
```

Cell 7: Tổng hợp toàn bộ Data Audit

```python

# CELL 7: DATA AUDIT SUMMARY

# Kiểm tra lại các điều kiện từ dữ liệu gốc
invoice_text = df["InvoiceNo"].astype("string").str.strip()
cancelled_mask = invoice_text.str.upper().str.startswith(
    "C", na=False
)

quantity = pd.to_numeric(df["Quantity"], errors="coerce")
price = pd.to_numeric(df["UnitPrice"], errors="coerce")

invoice_dates = pd.to_datetime(
    df["InvoiceDate"], errors="coerce"
)

# Tổng số dòng
total_rows = len(df)

# Tạo bảng tổng hợp
audit_summary = pd.DataFrame({
    "Metric": [
        "Total Rows",
        "Total Columns",
        "Missing CustomerID",
        "Missing Description",
        "Exact Duplicate Rows",
        "Cancelled Rows",
        "Cancelled Invoices",
        "Negative Quantity",
        "Zero Quantity",
        "Negative UnitPrice",
        "Zero UnitPrice",
        "Invalid InvoiceDate"
    ],
    "Value": [
        total_rows,
        df.shape[1],
        df["CustomerID"].isna().sum(),
        df["Description"].isna().sum(),
        df.duplicated().sum(),
        cancelled_mask.sum(),
        invoice_text[cancelled_mask].nunique(),
        (quantity < 0).sum(),
        (quantity == 0).sum(),
        (price < 0).sum(),
        (price == 0).sum(),
        invoice_dates.isna().sum()
    ]
})

# Hiển thị kết quả
print("===== DATA AUDIT SUMMARY =====")
display(audit_summary)
```

Phân loại giao dịch (Transaction Classification)

```python

# CELL 8: TRANSACTION CLASSIFICATION

import pandas as pd
import numpy as np

# 1. Tạo bản sao để không thay đổi dữ liệu gốc
work = df.copy()

# 2. Chuẩn hóa mã hóa đơn
work["InvoiceNo"] = (
    work["InvoiceNo"]
    .astype("string")
    .str.strip()
)

# 3. Chuẩn hóa ngày giao dịch
work["InvoiceDate"] = pd.to_datetime(
    work["InvoiceDate"],
    errors="coerce"
)

# 4. Đánh dấu hóa đơn hủy
work["InvoiceCancelled"] = (
    work["InvoiceNo"]
    .str.upper()
    .str.startswith("C", na=False)
)

# 5. Phân loại giao dịch
work["TransactionType"] = np.select(
    [
        work["InvoiceCancelled"],
        work["Quantity"] < 0,
        work["UnitPrice"] <= 0
    ],
    [
        "Cancelled",
        "Return_or_Negative_Quantity",
        "Zero_or_Negative_Price"
    ],
    default="Sale_Candidate"
)

# 6. Thống kê kết quả phân loại
print("===== TRANSACTION CLASSIFICATION =====")

display(
    work["TransactionType"].value_counts()
)

# 7. Kiểm tra số dòng
print("Số dòng dữ liệu gốc:", len(df))
print("Số dòng sau phân loại:", len(work))

# 8. Xem 10 dòng đầu tiên
display(
    work[
        [
            "InvoiceNo",
            "Quantity",
            "UnitPrice",
            "InvoiceCancelled",
            "TransactionType"
        ]
    ].head(10)
)
```

```python

# CELL 9: CLEAN SALES DATASET

# 1. Lọc các dòng bán hàng hợp lệ
sales = work.loc[
    (~work["InvoiceCancelled"]) &
    (work["Quantity"] > 0) &
    (work["UnitPrice"] > 0) &
    (work["InvoiceDate"].notna()) &
    (work["InvoiceNo"].notna()) &
    (work["InvoiceNo"] != "")
].copy()

# 2. Tạo cột doanh thu
sales["Revenue"] = (
    sales["Quantity"] * sales["UnitPrice"]
)

# 3. Tạo các cột thời gian
sales["Year"] = sales["InvoiceDate"].dt.year
sales["Month"] = sales["InvoiceDate"].dt.month

sales["YearMonth"] = (
    sales["InvoiceDate"]
    .dt.to_period("M")
    .astype(str)
)

# 4. Kiểm tra kết quả làm sạch
print("===== SALES DATA CLEANING =====")

print("Số dòng dữ liệu gốc:", len(df))
print("Số dòng bán hàng hợp lệ:", len(sales))
print("Số dòng không thuộc tập sales:",
      len(df) - len(sales))

# 5. Kiểm tra các điều kiện
print("\n===== VALIDATION =====")

print("Hóa đơn hủy còn lại:",
      sales["InvoiceCancelled"].sum())

print("Quantity <= 0:",
      (sales["Quantity"] <= 0).sum())

print("UnitPrice <= 0:",
      (sales["UnitPrice"] <= 0).sum())

print("InvoiceDate bị thiếu:",
      sales["InvoiceDate"].isna().sum())

# 6. Xem dữ liệu sau làm sạch
print("\n===== CLEANED DATA SAMPLE =====")
display(sales.head(10))
```

CELL 10: Tính các chỉ số kinh doanh (KPI)

```python
# CELL 10: BUSINESS KPI CALCULATION

# 1. Tổng doanh thu bán hàng hợp lệ
gross_sales = sales["Revenue"].sum()

# 2. Số hóa đơn duy nhất
orders = sales["InvoiceNo"].nunique()

# 3. Số khách hàng xác định được
identified_customers = sales["CustomerID"].nunique()

# 4. Tổng số lượng sản phẩm bán
units_sold = sales["Quantity"].sum()

# 5. Giá trị trung bình mỗi hóa đơn
aov = gross_sales / orders if orders > 0 else float("nan")

# 6. Hiển thị kết quả
print("===== BUSINESS KPI SUMMARY =====")

print(f"Gross Sales (GBP): £{gross_sales:,.2f}")
print(f"Orders: {orders:,}")
print(f"Identified Customers: {identified_customers:,}")
print(f"Units Sold: {units_sold:,}")
print(f"Average Order Value (GBP): £{aov:,.2f}")

# 7. Tạo bảng KPI tổng hợp
kpi_summary = pd.DataFrame({
    "KPI": [
        "Gross Sales (GBP)",
        "Orders",
        "Identified Customers",
        "Units Sold",
        "Average Order Value (GBP)"
    ],
    "Value": [
        round(gross_sales, 2),
        orders,
        identified_customers,
        units_sold,
        round(aov, 2)
    ]
})

display(kpi_summary)
```

Gross Sales £10,67 triệu: Tổng giá trị bán hàng hợp lệ trong toàn bộ giai đoạn dữ liệu, chưa trừ đầy đủ hoàn trả và điều chỉnh.



19.960 Orders: Có 19.960 hóa đơn bán hàng duy nhất.



4.338 Identified Customers: Có 4.338 khách hàng xác định được qua CustomerID.



5.588.376 Units Sold: Tổng số lượng sản phẩm được ghi nhận trong tập bán hàng.



AOV £534,40: Mỗi hóa đơn có giá trị trung bình khoảng £534,40.

CELL 11: Xuất dữ liệu đã làm sạch

```python

# CELL 11: EXPORT CLEANED DATASET

from pathlib import Path

# 1. Xác định thư mục gốc dự án
ROOT = Path.cwd()

if ROOT.name == "notebooks":
    ROOT = ROOT.parent

# 2. Tạo thư mục xuất dữ liệu
output_dir = ROOT / "data" / "processed"
output_dir.mkdir(parents=True, exist_ok=True)

# 3. Xuất dữ liệu bán hàng đã làm sạch
sales_path = output_dir / "sales_clean.csv"

sales.to_csv(
    sales_path,
    index=False,
    encoding="utf-8-sig"
)

# 4. Xuất bảng KPI
kpi_path = output_dir / "kpi_summary.csv"

kpi_summary.to_csv(
    kpi_path,
    index=False,
    encoding="utf-8-sig"
)

print("===== EXPORT RESULTS =====")
print("Sales dataset:", sales_path)
print("KPI summary:", kpi_path)

print("\nSố dòng đã xuất:", len(sales))
print("Số cột đã xuất:", sales.shape[1])

print("\nXuất dữ liệu thành công!")
```

```python

# CELL 12: VALIDATE EXPORTED DATA

import numpy as np

# Đọc lại file CSV
sales_check = pd.read_csv(
    sales_path,
    dtype={
        "InvoiceNo": "string",
        "StockCode": "string"
    },
    parse_dates=["InvoiceDate"]
)

print("===== EXPORT VALIDATION =====")

print("Số dòng CSV:", len(sales_check))

print("Gross Sales CSV:",
      round(sales_check["Revenue"].sum(), 2))

print("Orders CSV:",
      sales_check["InvoiceNo"].nunique())

# Kiểm tra tính nhất quán
assert len(sales_check) == len(sales)

assert np.isclose(
    sales_check["Revenue"].sum(),
    gross_sales,
    rtol=0,
    atol=0.00001
)

assert (
    sales_check["InvoiceNo"].nunique()
    == orders
)

print("\nVALIDATION PASSED!")
```
